```python
# Exemplo 01
# 1) Exemplos para substituição de acentuação: Mapa de Caracteres

def remover_acentos_substituicao(texto):
    mapa = {
        'á': 'a', 'à': 'a', 'â': 'a', 'ã': 'a', 'ä': 'a',
        'é': 'e', 'è': 'e', 'ê': 'e', 'ë': 'e',
        'í': 'i', 'ì': 'i', 'î': 'i', 'ï': 'i',
        'ó': 'o', 'ò': 'o', 'ô': 'o', 'õ': 'o', 'ö': 'o',
        'ú': 'u', 'ù': 'u', 'û': 'u', 'ü': 'u',
        'ç': 'c',
        'Á': 'A', 'À': 'A', 'Â': 'A', 'Ã': 'A', 'Ä': 'A',
        'É': 'E', 'È': 'E', 'Ê': 'E', 'Ë': 'E',
        'Í': 'I', 'Ì': 'I', 'Î': 'I', 'Ï': 'I',
        'Ó': 'O', 'Ò': 'O', 'Ô': 'O', 'Õ': 'O', 'Ö': 'O',
        'Ú': 'U', 'Ù': 'U', 'Û': 'U', 'Ü': 'U',
        'Ç': 'C'
    }

    return ''.join(mapa.get(c, c) for c in texto)
```

```python
# Exemplo 01
# 1) Exemplos para substituição de acentuação: Mapa de Caracteres

texto_original='João da Conceição'

texto_tratado = remover_acentos_substituicao(texto_original)  

print(texto_tratado.upper()) 
```

```python
# Exemplo 02
# Exemplo para substituição de acentuação: Normalização Unicode (forma NFD)

import unicodedata
import re

def remover_acentos_unicode(texto):
    texto_normalizado = unicodedata.normalize('NFD', texto)
    texto_sem_acentos = re.sub(r'[\u0300-\u036f]', '', texto_normalizado)
    return texto_sem_acentos
```

```python
# Exemplo 02
# Exemplo para substituição de acentuação: Normalização Unicode (forma NFD)

texto_original = 'André Álvares João'

texto_tratado = remover_acentos_unicode(texto_original)  

print(texto_tratado.upper()) 
```

```python
# Exemplo 04
# Padronização de Espaços e Caractere Especiais
# Exemplos para normalização e comparação de nomes: Expansão de Iniciais
apelidos = {
    "V": "Vicente",
    "J": "José",
    "M": "Maria"
}

def expandir_iniciais(nome, mapa):
    palavras = nome.split()
    novas = []
    for p in palavras:
        if len(p) == 1 and p in mapa:
            novas.append(mapa[p])
        else:
            novas.append(p)
    return " ".join(novas)
```

```python
# Exemplo 04
# Padronização de Espaços e Caractere Especiais
# Exemplos para normalização e comparação de nomes: Expansão de Iniciais

nome = 'Marcelo V Rosim'

nome_exp = expandir_iniciais(nome, apelidos).upper()

print(nome_exp)  # Marcelo Vicente Rosim
```

```python
# Exemplo 05
# Exemplos para normalização e comparação de nomes: Distância de Levenshtein

import Levenshtein

nome1 = "Marcelo Rosim"
nome2 = "Marcelo Vicente Rosim"

distancia = Levenshtein.distance(nome1, nome2)
similaridade = Levenshtein.ratio(nome1, nome2)

print(f"Distância: {distancia}")
print(f"Similaridade: {round(similaridade * 100)}%")
```

```python
# Exemplo 07
# Nomes com escritas bem diferentes, mas som igual
nome_a = "Stephen"
nome_b = "Steven"

cod_a = jellyfish.metaphone(nome_a)
cod_b = jellyfish.metaphone(nome_b)

print(f"Metaphone {nome_a}: {cod_a}") # Retorna STFN
print(f"Metaphone {nome_b}: {cod_b}") # Retorna STFN
```

```python
import jellyfish

def verificar_similaridade(string1, string2):
    # O Soundex de "Luciano" é L250
    # O Soundex de "Luciana" é L250
    codigo1 = jellyfish.soundex(string1)
    codigo2 = jellyfish.soundex(string2)

    if codigo1 == codigo2:
        return f"SIMILAR (Ambos geraram o código {codigo1})"
    else:
        return f"DIFERENTE ({codigo1} vs {codigo2})"

# Testando
print(verificar_similaridade("Luciano", "Luciana")) 
# Retorno: SIMILAR (Ambos geraram o código L250)
```
